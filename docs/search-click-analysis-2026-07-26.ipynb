# ACT Creative search and click performance

## tl;dr

- Google Search Console clicks rose from 29 to 85 (+193.1%) and impressions rose from 3,194 to 7,593 (+137.7%).
- Vercel recorded 108 visitors (+248%) and 155 page views (+80%) over the nearly aligned 30-day window.
- Growth is real, but engagement quality is the main weakness: bounce rate reached 80% (+15%).
- The clearest SEO opportunities are booth design/build, event fabrication, venue pages, and custom props.

## Context & Methods

This companion notebook records the reviewed dashboard values used in the report. Search Console covers 2026-06-26 through 2026-07-23 and compares the prior 28 days. Vercel covers 2026-06-26 through 2026-07-26 and compares the prior 30 days. The three-day end-date difference reflects Search Console reporting latency.

### Key Assumptions

- Search Console clicks are interaction counts, while Vercel visitors are privacy-preserving unique visitors; they are not expected to reconcile one-to-one.
- Vercel percentage deltas are taken directly from the dashboard. Prior-period absolute values are estimates derived from those rounded deltas.
- Ranking improvement is shown as a reduction in average position; a smaller position number is better.

## Data

```python
gsc = {
    "clicks_current": 85,
    "clicks_prior": 29,
    "impressions_current": 7593,
    "impressions_prior": 3194,
    "ctr_current": 0.011,
    "ctr_prior": 0.009,
    "position_current": 27.7,
    "position_prior": 48.2,
}

vercel = {
    "visitors": 108,
    "visitors_change": 2.48,
    "pageviews": 155,
    "pageviews_change": 0.80,
    "bounce_rate": 0.80,
    "bounce_change": 0.15,
    "google_visitors": 39,
    "google_pageviews": 47,
    "homepage_visitors": 74,
    "homepage_pageviews": 101,
}

landing_pages = [
    {"page": "Food truck rental", "clicks": 12, "prior_clicks": 3, "impressions": 388, "ctr": 0.031, "position": 16.5},
    {"page": "Homepage", "clicks": 12, "prior_clicks": 17, "impressions": 70, "ctr": 0.171, "position": 4.7},
    {"page": "Event fabrication Hong Kong", "clicks": 9, "prior_clicks": 2, "impressions": 105, "ctr": 0.086, "position": 22.6},
    {"page": "Booth design/build Singapore", "clicks": 8, "prior_clicks": 0, "impressions": 2004, "ctr": 0.004, "position": 46.8},
    {"page": "Custom props Singapore", "clicks": 7, "prior_clicks": 3, "impressions": 435, "ctr": 0.016, "position": 30.6},
    {"page": "Event fabrication Singapore", "clicks": 5, "prior_clicks": 1, "impressions": 862, "ctr": 0.006, "position": 14.1},
    {"page": "FRP sculpture fabrication", "clicks": 5, "prior_clicks": 0, "impressions": 63, "ctr": 0.079, "position": 7.8},
    {"page": "Event venue finder", "clicks": 3, "prior_clicks": 1, "impressions": 245, "ctr": 0.012, "position": 7.8},
    {"page": "National Museum venue", "clicks": 3, "prior_clicks": 0, "impressions": 229, "ctr": 0.013, "position": 7.9},
    {"page": "Marina Bay Sands venue", "clicks": 2, "prior_clicks": 0, "impressions": 381, "ctr": 0.005, "position": 11.8},
]
```

## Results

```python
def relative_change(current, prior):
    return current / prior - 1

results = {
    "gsc_click_change": relative_change(gsc["clicks_current"], gsc["clicks_prior"]),
    "gsc_impression_change": relative_change(gsc["impressions_current"], gsc["impressions_prior"]),
    "gsc_ctr_relative_change": relative_change(gsc["ctr_current"], gsc["ctr_prior"]),
    "gsc_position_improvement": gsc["position_prior"] - gsc["position_current"],
    "google_share_of_visitors": vercel["google_visitors"] / vercel["visitors"],
    "homepage_share_of_visitors": vercel["homepage_visitors"] / vercel["visitors"],
    "top_page_growth_contribution": sum(p["clicks"] - p["prior_clicks"] for p in landing_pages),
}

for key, value in results.items():
    print(f"{key}: {value:.3f}")
```

```python
opportunities = []
for page in landing_pages:
    if page["impressions"] >= 200 and page["ctr"] < 0.02:
        opportunities.append({
            "page": page["page"],
            "impressions": page["impressions"],
            "ctr": page["ctr"],
            "position": page["position"],
        })

for row in sorted(opportunities, key=lambda x: x["impressions"], reverse=True):
    print(row)
```

## Takeaways

1. Visibility and search demand expanded faster than click efficiency, so title/meta and intent matching are now the highest-leverage SEO work.
2. Booth design/build has the largest near-term volume opportunity: 2,004 impressions but only 0.4% CTR.
3. Venue content already ranks near page one, but CTR is low; richer snippets and stronger intent alignment should be tested.
4. Vercel traffic grew strongly, but the 80% bounce rate means conversion and on-page engagement need to be instrumented before declaring the traffic fully successful.
